```python
# Create a structural inventory of the PUMF variables

variable_inventory = []

for col in columns:
    prefix = col[:3]

    # Fourth character indicates variable type for standard CCHS names
    if len(col) >= 4:
        var_type_code = col[3]
    else:
        var_type_code = ""

    type_meaning = {
        "_": "Collected",
        "C": "Coded",
        "D": "Derived",
        "F": "Flag",
        "G": "Grouped"
    }.get(var_type_code, "Other / special")

    variable_inventory.append({
        "Variable": col,
        "Module": prefix,
        "Type_Code": var_type_code,
        "Type": type_meaning
    })

variable_inventory = pd.DataFrame(variable_inventory)

display(variable_inventory)
```

## CCHS PUMF Exploration — Stage 3: Special Codes and Survey Missingness

## What I investigated

I examined the frequency of commonly occurring coded values, particularly `6` and `9`, across the CCHS PUMF.

This was done after finding that the dataset contains no literal `NaN` values.

## Important finding

The absence of `NaN` values does NOT mean that all variables contain complete respondent information.

The CCHS PUMF uses coded response values for different situations, including:

- valid responses,
- valid skips / not-applicable cases,
- and "Not stated" responses.

The meaning of a numerical code is **variable-specific** and therefore cannot be interpreted globally.

For example:

- `ECV2_10` contains a very large number of code `6` values. The official CCHS data dictionary identifies these as **Valid skip**, because the question applies only to a specific subset of respondents.
- `LSM_01` contains many respondents with code `09`, but `09` is a legitimate answer on its 0–10 life-satisfaction scale. The "Not stated" value for this variable is `99`.
- `INCDGRRS` contains code `9` values that are explicitly identified as **Not stated**.

Therefore, a global transformation such as "replace every 6 or 9 with missing" would be incorrect.

## Methodological implication

Before preparing variables for machine learning, special codes must be interpreted using the official CCHS documentation and the universe of each variable.

This means that apparent missingness has to be distinguished into at least:

1. Valid response
2. Valid skip / not applicable
3. Not stated / non-response
4. Other variable-specific special codes

## CCHS PUMF Exploration — Stage 5: Module Applicability and Survey Universes

## What I investigated

I used the official CCHS 2022 data dictionary to understand how the `DO...` inclusion flags relate to the variables within each module.

A module-level flag tells us whether a section of the questionnaire is applicable or available for a respondent, but individual variables can also have their own survey universe.

## Key finding

The module-availability patterns observed in Stage 4 are therefore not ordinary missing values.

For example, some variables are only applicable to particular age groups, geographic groups, or respondents who followed a specific path through the questionnaire. When a variable contains `6 = Valid skip`, this can mean that the respondent was outside its universe rather than that information was missing.

The same principle explains why some modules are available only for the territorial respondents while others have the opposite pattern.

## Methodological implication

I should not treat every valid skip as missing and then impute it.

Before using a variable as a feature, I need to check:

- who the variable applies to,
- what its valid-skip codes mean,
- whether its universe is compatible with the intended modelling population,
- and whether the variable can be compared consistently across respondents.

This means that feature availability is part of feature selection, not just a preprocessing issue.

## Research implication

The original 11-feature model should not be treated as the full information available in the CCHS PUMF. However, broader feature sets need to be constructed around variables whose definitions and applicable populations are understood.

Variables with restricted universes can instead be tested in separate population-specific experiments rather than being forced into a single full-sample model.

### How the decisions were made

Variables from the A sections were considered first because they provided broadly applicable demographic, socioeconomic, anthropometric, behavioural, and contextual information. Where several variables represented similar information, a more compact representation was preferred. For example, `ALCDVTTM` was used instead of retaining multiple overlapping alcohol measures, and `HWTDGISW` and `HWTDGWHO` were retained together because they represent the adult and 12–17 BMI classifications needed for the full 12+ population.

The B-section review then considered variables that were scientifically plausible but required additional caution because of age restrictions, conditional question paths, overlap, or specialised interpretation.

`ECV_05` was selected as the broadly applicable vaping representation, while the more detailed vaping variables were reserved for later analysis. `SMKDVSTY` was not added to the main 12+ model because its universe is restricted to adults and it will instead be considered in an adult-only smoking experiment. Labour-force and main-activity variables were also reserved for working-age experiments. `WTP_50` was held out because it provides subjective weight perception and may be better evaluated separately against the selected BMI representations.

The three social-demographic variables were treated separately: `SDCDGIMM` was included as the initial contextual representation, while `SDCDVABT` and `SDCDVFLA` were reserved for a later demographic experiment.

`RHC_05` was held out because healthcare-provider use may reflect healthcare access or diagnosis-related processes and therefore represents a different type of information from the main feature set.

### Health-condition and comorbidity decision

Other diagnosed health-condition variables, including high blood pressure, high cholesterol, and cardiovascular conditions, were not automatically added to this preliminary set. They may provide substantial predictive information, but including them changes the information available to the model by incorporating existing health-condition history.

They will therefore be evaluated separately in an expanded health-context experiment after their definitions and applicability have been verified. This allows their incremental contribution to be tested rather than assuming in advance that they belong in the main model.

### Research interpretation

The 12 variables are a **preliminary CCHS documentation- and domain-guided feature set**, not a claim that these are the universally optimal predictors of diabetes.

The selections were made before model performance was evaluated. The purpose of the next modelling stage is therefore to test how this research-guided representation performs relative to the original baseline and, later, to more data-driven feature-selection approaches.

The Core, Review, and Restricted categories used earlier are working research classifications created for this project rather than official Statistics Canada classifications.

## Stage 15 — Literature-Informed Feature Verification

The preliminary feature set was constructed using CCHS documentation and domain reasoning. The literature review was then used to verify which predictor concepts had stronger support for diabetes classification and whether appropriate CCHS 2022 variables were available to represent them.

For each literature-supported predictor, I verified:

- what the CCHS variable measures;
- who it applies to;
- how its response categories are coded;
- which values represent valid skips or not-stated responses;
- whether it is applicable to the intended modelling population; and
- whether it belongs in the main feature set or a separate experiment.

The purpose of this stage is to connect the literature evidence to the actual CCHS variables rather than assuming that every literature-supported predictor can be used directly in the model.

### Stage 15.1 — BMI / Body-Weight Measures

BMI was identified in the literature review as an important predictor of diabetes classification. The CCHS 2022 PUMF provides three relevant grouped BMI measures:

- `HWTDGISW`: BMI classification for adults aged 18+, based on self-reported height and weight using the international classification.
- `HWTDGWHO`: BMI classification for respondents aged 12–17, based on self-reported height and weight using WHO age- and sex-specific classification.
- `HWTDGBCC`: adjusted BMI classification for adults aged 18+.

The first two variables have different age-specific universes and therefore cannot be treated as interchangeable representations of the same population.

### BMI Feature Decision

For the literature-informed feature set, `HWTDGISW` and `HWTDGWHO` will be retained because together they provide BMI information across the full modelling population aged 12 and older while respecting the CCHS age-specific definitions.

`HWTDGBCC` will be reserved for a controlled experiment comparing the adult self-reported BMI representation with the adjusted adult representation rather than including both in the same feature set.

CCHS special codes will be handled according to the variable documentation. In particular, code 6 represents a population exclusion or valid skip, while code 9 represents a not-stated response. These values will not be treated as ordinary missing observations.

### Stage 15.2 — Hypertension and Other Health-Condition Variables

The literature review identified hypertension and other diagnosed health conditions as important predictor families that were under-represented in the preliminary feature set. Canadian diabetes risk-prediction work, including DPoRT, has used hypertension as a predictor.

The relevant CCHS 2022 variables were then checked for their definitions, universes, response coding, and suitability for controlled modelling experiments.

### Verified Health-Condition Candidates

| Variable | What it measures | Universe | Coding | Preliminary decision |
|---|---|---|---|---|
| `CCC_80` | Has high blood pressure | Respondents with `DOCCC=1` | 1 Yes, 2 No, 9 Not stated | High-priority clinical experiment |
| `CCC_85` | Took medication for high blood pressure in past month | Respondents with `DOCCC=1` | 1 Yes, 2 No, 9 Not stated | Hold; related to `CCC_80` |
| `CCC_90` | Has ever been told by a health professional that blood cholesterol was high | Respondents with `DOCCC=1` | 1 Yes, 2 No, 9 Not stated | Expanded health-context experiment |
| `CCCDGCAR` | Heart disease, heart attack, or effects of stroke | Respondents aged 35+ with `DOCCC=1` | 1 Yes, 2 No, 6 Valid skip, 9 Not stated | Age-restricted secondary experiment |

### Health-Condition Feature Decision

These variables will be treated as a separate clinical information block rather than being added indiscriminately to the main feature set.

`CCC_80` is the highest-priority candidate because hypertension has established support in diabetes prediction literature and the CCHS variable provides a directly defined measure of reported high blood pressure.

`CCC_90` will be evaluated in an expanded clinical-context experiment. It represents whether a health professional has previously told the respondent that their blood cholesterol was high, rather than providing a laboratory measurement.

`CCC_85` will be held out initially because recent antihypertensive medication use is closely related to hypertension status and introduces treatment-related information. Its incremental contribution can be examined after the contribution of `CCC_80` has been established.

`CCCDGCAR` will be reserved for a separate experiment among respondents aged 35 and older because its CCHS universe is age-restricted. Its valid-skip values will not be treated as evidence that a younger respondent does not have cardiovascular disease.

The purpose of these experiments is to test whether clinically reported health information provides incremental predictive value beyond the demographic, anthropometric, socioeconomic, behavioural, and contextual information in the primary feature representation.

### Stage 15.3 — Physical Activity

Physical activity was identified in the literature review as an important predictor of diabetes classification. The CCHS 2022 PUMF contains several derived physical-activity measures, but their applicability must be considered before using them as predictors.

### CCHS Physical-Activity Applicability

The PAA module is identified by the inclusion flag `DOPAA`. In this PUMF, 1,949 respondents have `DOPAA = 1`, while 65,130 have `DOPAA = 0`.

The detailed PAA variables apply to respondents aged 18 and older for whom the module is available. They therefore cannot be used directly in the primary modelling population of respondents aged 12 and older.

These structural skips do not indicate zero physical activity or ordinary missing observations. They reflect the applicability of the module to the respondent.

### Main Physical-Activity Measures Identified

Several derived variables represent related aspects of physical activity:

- `PAADVMVA`: total minutes of moderate-to-vigorous physical activity in the previous 7 days.
- `PAADVACV`: physical-activity indicator based on the Canadian Physical Activity Guidelines.
- `PAADVAC2`: four-level physical-activity classification.
- `PAADVVOL`: weekly physical-activity volume in MET-minutes/week.
- `PAADVWHO`: physical-activity classification based on WHO guidelines.

These measures will not automatically be combined because several represent different transformations of the same underlying activity information.

### Physical-Activity Feature Decision

Physical activity will not be included in the primary all-age 12+ feature set because the PAA module is restricted in both age and availability.

Instead, it will be investigated in a restricted adult PAA experiment using respondents for whom the module is available. The purpose will be to determine whether physical-activity information provides incremental predictive value beyond the established feature set.

`PAADVMVA` will be used as the initial candidate representation because it retains the reported amount of moderate-to-vigorous activity rather than immediately reducing it to a categorical measure. Alternative derived representations can be compared in a later controlled experiment if there is a methodological reason to do so.

This preserves the distinction between scientific relevance and applicability: physical activity may be an important predictor, but its CCHS 2022 availability makes it inappropriate for automatic inclusion in the primary 12+ model.

### Stage 15.4 — Smoking

Smoking was identified in the literature review as a more established diabetes-prediction variable than vaping. The CCHS 2022 PUMF contains several smoking-related measures, but their population applicability and overlap need to be considered before selecting a representation.

### Applicability of `SMKDVSTY`

`SMKDVSTY`, used in the original baseline, represents smoking status for adult respondents. Because its universe is restricted to adults, it cannot be included directly in the primary modelling population of respondents aged 12 and older.

Age-based valid skips should therefore not be interpreted as evidence of non-smoking.

### Smoking Feature Decision

`SMKDVSTY` will be reserved for a separate adult-only smoking experiment. This experiment will test whether smoking information provides incremental predictive value after accounting for the established feature set.

The CCHS smoking section also contains separate measures of current smoking, past smoking, and vaping. These will not be combined indiscriminately. The smoking experiment will begin with a clearly defined smoking-status representation so that any performance change can be attributed to the addition of smoking information rather than a collection of overlapping variables.

The literature review provides stronger support for smoking than for simple ever-vaping status. Therefore, `SMKDVSTY` will be treated as the primary smoking candidate for the adult experiment, while `ECV_05` will remain a secondary or exploratory predictor rather than a core feature.

### Stage 15.5 — Family History and Waist Circumference

The literature review identified family history of diabetes and waist circumference as important predictors used in diabetes prediction models. These predictors were therefore checked against the CCHS 2022 PUMF before being considered for modelling.

The key question was whether the PUMF contains a clearly identifiable variable that represents each predictor and is suitable for the intended modelling population.

### Family History of Diabetes

Family history has strong support in the diabetes prediction literature and is included in established diabetes risk-scoring approaches.

However, an exact family-history-of-diabetes variable could not be sufficiently verified in the CCHS 2022 PUMF documentation reviewed for this project. No variable will therefore be assigned to this predictor without supporting CCHS documentation.

**Decision:** Family history is recorded as a literature-supported predictor that could not be sufficiently verified in the analytical CCHS 2022 PUMF and will not be included in the current feature sets.

### Waist Circumference

Waist circumference is also widely used in diabetes prediction models and represents a different aspect of body composition from BMI.

An appropriate waist-circumference variable could not be sufficiently verified in the CCHS 2022 PUMF documentation reviewed for this project. The available BMI variables will therefore not be treated as substitutes for waist circumference.

**Decision:** Waist circumference is recorded as a literature-supported predictor that could not be sufficiently verified in the analytical CCHS 2022 PUMF and will not be included in the current feature sets.

### Research Limitation

These exclusions reflect dataset availability and variable verification, not a conclusion that the predictors are unimportant.

This distinction is maintained throughout the feature-selection process: literature evidence, CCHS availability, population applicability, and modelling suitability are treated as separate criteria.

### Stage 15.6 — Cholesterol and Cardiovascular Disease

The literature review identified high cholesterol and cardiovascular disease as clinically relevant predictor families. The CCHS 2022 PUMF contains variables representing both concepts, but their exact definitions and population restrictions need to be considered before modelling.

### `CCC_90` — High Cholesterol

`CCC_90` represents whether the respondent has ever been told by a health professional that their blood cholesterol was high.

Its CCHS universe is respondents with `DOCCC = 1`, with the following response codes:

- `1` = Yes
- `2` = No
- `9` = Not stated

This is a reported history of high cholesterol rather than a measured laboratory cholesterol value.

### `CCCDGCAR` — Cardiovascular Condition

`CCCDGCAR` is a grouped cardiovascular-condition variable covering heart disease, previous heart attack, or effects of stroke.

Its universe is restricted to respondents aged 35 and older with `DOCCC = 1`. The response codes are:

- `1` = Yes
- `2` = No
- `6` = Valid skip
- `9` = Not stated

Because of this age restriction, a valid skip cannot be interpreted as evidence that a younger respondent does not have cardiovascular disease.

### Clinical-Context Feature Decision

`CCC_90` will be evaluated in the expanded clinical-context feature set. Its contribution will be interpreted as information from previously reported health-condition history rather than as a laboratory measurement.

`CCCDGCAR` will be evaluated separately among respondents aged 35 and older because its restricted universe makes it unsuitable for the primary all-age model.

The clinical experiments will therefore test whether these health-condition variables provide incremental predictive information beyond the demographic, anthropometric, socioeconomic, behavioural, and contextual features in the primary representation. Because the CCHS 2022 data are cross-sectional, the modelling results will not establish a temporal or causal relationship between these conditions and diabetes status.

## 15.7 — Social-Demographic Representation

The literature review identified immigration- and identity-related characteristics as potentially relevant predictors in diabetes classification and risk models. These concepts are not interchangeable, so the corresponding CCHS variables were examined separately.

### 15.7.1 Relevant CCHS 2022 Variables

The CCHS 2022 PUMF contains:

- `SDCDGIMM`: immigrant flag
- `SDCDVABT`: Indigenous identity
- `SDCDVFLA`: visible minority flag

`SDCDGIMM` distinguishes non-immigrant respondents from immigrant or non-permanent resident categories.

`SDCDVABT` identifies respondents reporting Indigenous identity, including First Nations, Métis, or Inuk/Inuit identity.

`SDCDVFLA` identifies respondents classified as a visible minority.

These variables represent different social-demographic characteristics and are therefore not treated as interchangeable predictors.

## 15.10 — Modelling Protocol

The modelling stage will use a consistent protocol so that differences between experiments can be attributed primarily to the feature condition and model rather than changes in preprocessing or evaluation.

For each experiment:

1. select a frozen feature condition;
2. define the corresponding eligible modelling population;
3. exclude unusable target responses;
4. create a fixed stratified train/validation/test split;
5. fit preprocessing using the training data only;
6. fit the model using the training data;
7. use the validation set for model-selection and threshold decisions;
8. lock the selected configuration and threshold;
9. evaluate the final model once on the untouched test set; and
10. record performance, interpretation, and failure cases.

The test set will not be used for feature selection, hyperparameter tuning, or threshold selection.

### 15.10.1 Data Splitting

For experiments using the same modelling population, the same fixed stratified split will be retained so that feature-set and model comparisons are made on the same observations.

Population-restricted experiments, such as the 35+ clinical experiment and restricted adult behavioural experiments, will use a fixed stratified split generated within their eligible population.

The training set is used to learn model parameters and preprocessing parameters.

The validation set is used for model-selection decisions, threshold selection, and other predefined modelling choices.

The final test set is reserved for locked evaluation.

### 15.10.2 Preprocessing

All preprocessing operations must be fitted using the training data only.

Categorical variables will be encoded using an appropriate categorical encoding strategy, while numerical variables will be processed according to their data type and missing-value structure.

The fitted preprocessing pipeline will then be applied unchanged to the validation and test sets.

CCHS-specific special codes will be interpreted using variable-specific documentation rather than a global missing-value rule.

### 15.10.3 Class Imbalance

Diabetes is the minority target class.

Where appropriate, class-weighted training will be used to increase the influence of minority-class observations during model fitting. The class-weighting strategy will be recorded for each experiment.

Performance will not be evaluated using accuracy alone because accuracy can be misleading when the target classes are imbalanced.

### 15.10.4 Models

The primary models will be:

1. Logistic Regression
2. Random Forest
3. Gradient Boosting

These models provide complementary modelling perspectives:

- Logistic Regression provides an interpretable linear reference model.
- Random Forest provides a nonlinear ensemble model.
- Gradient Boosting provides a second nonlinear ensemble approach.

Additional algorithms will not be added solely to increase the number of models. Any additional model must have a methodological reason for inclusion.

### 15.10.5 Evaluation Metrics

Each experiment will report:

- Accuracy
- Precision
- Recall
- F1-score
- ROC-AUC
- PR-AUC
- Confusion matrix

ROC-AUC will be used to assess threshold-independent discrimination.

PR-AUC will receive particular attention because the diabetes class is relatively uncommon.

Threshold-dependent metrics will be reported using the classification threshold selected from the validation set.

### 15.10.6 Threshold Selection

The default probability threshold of 0.50 will not automatically be assumed to be optimal.

For each model, the classification threshold will be selected on the validation set using a predefined evaluation rule.

Once selected, the threshold will be locked before final test evaluation.

The test set will not be examined when choosing the threshold.

### 15.10.7 Failure Analysis

For each final model, both successful and unsuccessful predictions will be examined.

Particular attention will be given to:

- false positives;
- false negatives;
- differences between training and validation/test performance;
- cases where different models disagree; and
- feature patterns associated with difficult cases.

Failure analysis will be used to understand why a model succeeds or fails rather than reporting performance metrics without interpretation.

### 15.10.8 Interpretation

Interpretation will depend on the model type.

For Logistic Regression, coefficient patterns will be examined within the one-hot encoded feature representation.

For tree-based models, feature importance or permutation importance may be used to examine which features contribute most strongly to predictions.

These measures will be interpreted as model-specific predictive information rather than causal evidence.

### 15.10.9 Survey Weight Limitation

The CCHS sampling weight `WTS_M` will not be treated as an ordinary predictor.

The primary machine-learning results therefore represent respondent-level predictive modelling rather than survey-weighted population estimates.

Any future weighted analysis will be treated as a separate sensitivity analysis with its own methodology.

### 15.10.10 Experiment Record

Each experiment will record:

- feature set;
- modelling population;
- model;
- preprocessing;
- class-weighting strategy;
- model configuration;
- threshold-selection rule;
- selected threshold;
- training performance;
- validation performance;
- final test performance;
- confusion matrix;
- feature interpretation;
- failure cases; and
- limitations.

This record will make it possible to explain what was tested, what changed between conditions, what improved or failed, and why the observed result changed.
